# NLP Scenario 
You are an analyst for a marketing company that just launched a new product suite of mobile devices. You have data from product reviews of the new product. You want to generate a product description based on what customers like most about the device, using a standard marketing template.

#####  Product Reviews 
1. "I absolutely love the TechWave X1! It has made my daily tasks so much easier and more efficient. Highly recommend it!"
2. "I'm not very impressed with the TechWave X1. It lacks some essential features and is quite slow."
3. "The TechWave X1 is fantastic! It has exceeded my expectations and has become an essential part of my daily routine."
4. "I found the TechWave X1 to be quite average. It does the job, but there's nothing particularly special about it."
5. "The TechWave X1 is terrible. It's full of glitches and crashes frequently. I regret purchasing it."
6.  "The TechWave X1 is disappointing. It doesn't live up to the hype and is missing several key functionalities."
   
#### Marketing template 

 "Introducing the TechWaveX1, the perfect device for...."

````python
# data setup 
# create a list of reviews 
reviews = ["I absolutely love the TechWave X1! It has made my daily tasks so much easier and more efficient. Highly recommend it!",
"I'm not very impressed with the TechWave X1. It lacks some essential features and is quite slow.",
"The TechWave X1 is fantastic! It has exceeded my expectations and has become an essential part of my daily routine.",
"I found the TechWave X1 to be quite average. It does the job, but there's nothing particularly special about it.",
"The TechWave X1 is terrible. It's full of glitches and crashes frequently. I regret purchasing it.",
"The TechWave X1 is disappointing. It doesn't live up to the hype and is missing several key functionalities."
]

#marketing template 
marketing_template =  "Introducing the TechWaveX1, the perfect device for...."
````

# Initialize the Text Generation Pipeline

The default text generation model can be called with:
```python
pipeline("text-generation")
```

Key generation parameters:
- max_length: Maximum length of generated text
- num_return_sequences: Number of different generations to return

````python
# initalize the text generation pipeline
text_writer = pipeline("text-generation", model="gpt2")
````

Whilst it downloads you could look at the [model card for GPT2](https://huggingface.co/openai-community/gpt2)

````python
# write a simple prompt 
prompt = "Write a product description for the TechWave X1"
````

````python
# generate basic completion using the prompt
result = text_writer(prompt, 
                  max_length=100,
                  num_return_sequences=1)
result
````

````python
print(result[0]["generated_text"])
````

# Prompting
Let's modify our prompt to include our reviews.

````python
# write a more detailed prompt
better_prompt = f"""Write a product description for the TechWave X1 that highlights its key features and benefits. 
         Using {marketing_template} as a starting point and base the description on the following reviews: {reviews}
        output a short description of the product that can be used in marketing materials.
"""
````

````python
# generate basic completion using the prompt
new_result = text_writer(better_prompt, 
                  max_length=500,
                  num_return_sequences=1)
````

````python
print(new_result[0]['generated_text'])
````
