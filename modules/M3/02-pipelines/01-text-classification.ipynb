# NLP Scenario 
You are an analyst for a marketing company that just launched a new product suite of mobile devices. You have data from product reviews of the new product. We need to draft a report for our client that reports the sentiment of the reviews.

````python
# data setup 
# create a list of reviews 
reviews = ["I absolutely love the TechWave X1! It has made my daily tasks so much easier and more efficient. Highly recommend it!",
"I'm not very impressed with the TechWave X1. It lacks some essential features and is quite slow.",
"The TechWave X1 is fantastic! It has exceeded my expectations and has become an essential part of my daily routine.",
"I found the TechWave X1 to be quite average. It does the job, but there's nothing particularly special about it.",
"The TechWave X1 is terrible. It's full of glitches and crashes frequently. I regret purchasing it.",
"The TechWave X1 is disappointing. It doesn't live up to the hype and is missing several key functionalities."
]
````

# Initialize the pipeline for your task: Sentiment Analysis 

The default analyzer for sentiment analysis can be called with: 
```python 
    pipeline("sentiment-analysis")
```
This pipeline can take data at sentence, paragraph, or document level and will return a sentiment score and a label in the format of:
```python 
 [{'score': 0.9991, 'label': 'POSITIVE'}]
```

The score represents the model's confidence in its prediction. A score closer to 1.0 (or 100%) indicates higher confidence in the predicted sentiment label so that learners going through it on their own can interpret the scores more easily.

Let's get started

````python
# initialize a sentiment analysis pipeline with default model
sentiment_analyzer = pipeline("sentiment-analysis")
````

````python
# now let's run this on a single review
sentiment_analyzer("I absolutely love the TechWave X1! It has made my daily tasks so much easier and more efficient. Highly recommend it!")
````

#### Which model is used by default in the pipeline for sentiment analysis?
In our case it defaulted to [distilbert-base](https://huggingface.co/distilbert/distilbert-base-uncased-finetuned-sst-2-english)

# Finding a suitable model

One way to find suitable models in Hugging Face is to search:
- Go to the Hugging Face home page
- Click on **Models** at the top
- On the left hand side select **Tasks**
- Select **Text Classification** lower down on the left hand side
- Then search for ‘**sentiment**’ in the search box on the right

Suitable models will appear (usually thousands), and you can sort by ‘**Most likes**’.  Model size will give you an indication of how resource intensive the model will be.

Let's use the `nlptown/bert-base-multilingual-uncased-sentiment` model as it is ranked very highly.  (If you search within the results for the `roberta-large-mnli` model, it's not there.)

# Try It!

Now let's use a model that is fine-tuned specifically for sentiment analysis.

1. Load our `reviews.csv` file
2. Build a sentiment pipeline that uses [nlptown/bert-base-multilingual-uncased-sentiment](https://huggingface.co/nlptown/bert-base-multilingual-uncased-sentiment)
3. Compare the actual 5 star reviews with the predicted reviews in the data set

````python
# read in data (../data/reviews.csv) to a pandas DataFrame 
df = pd.read_csv('../data/reviews.csv')
````

````python
# initialize a sentiment analysis pipeline with your model
pipe = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment")
````

````python
%%time
# run the pipeline on the reviews column 
# note: our pipeline can take a list of documents, so .tolist() may be helpful
results = pipe(df["review"].tolist())
````

````python
# save your results as predicted_rating and predicted_score columns in the df
df[["predicted_rating", "predicted_score"]] = pd.DataFrame(results)  
# drop the star from the predicted_star column using df["predicted_star"].apply(lambda x: int(x.split(" ")[0])) and save it back to the column
df["predicted_rating"] = df["predicted_rating"].apply(lambda x: int(x.split(" ")[0]))
df.head()
````

````python
# check similarity between the predicted rating and the actual rating column 

# generate classification report
print("\nClassification Report:")
print(classification_report(df['rating'], df['predicted_rating']))
````

````python
pd.crosstab(df['rating'], df['predicted_rating'])
````
