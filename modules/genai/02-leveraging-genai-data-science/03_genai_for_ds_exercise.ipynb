# Using a chat model from code

Add a key for Gemini or OpenRouter, then run the setup cells.


```python
# Add the key for the provider you want to use.
GOOGLE_API_KEY = "YOUR_KEY_HERE"
OPENROUTER_API_KEY = "YOUR_KEY_HERE"

# Pick one provider for the notebook.
MODEL_PROVIDER = "gemini"  # or "openrouter"
GEMINI_MODEL = "models/gemini-2.5-flash"
OPENROUTER_MODEL = "nvidia/nemotron-3.5-lightning:free"
```

If a package import fails, run the install cell again before continuing.


```python
import google.generativeai as genai
from openai import OpenAI


def start_gemini_chat(api_key, model_name="models/gemini-2.5-flash"):
    # Create one Gemini chat session that we can reuse.
    genai.configure(api_key=api_key)
    model = genai.GenerativeModel(model_name)
    return model.start_chat(history=[])


def ask_gemini(chat, prompt):
    # Send one prompt and return only the text reply.
    return chat.send_message(prompt).text.strip()


def start_openrouter_client(api_key, model_name="nvidia/nemotron-3.5-lightning:free"):
    # OpenRouter uses the OpenAI client with a different base URL.
    client = OpenAI(base_url="https://openrouter.ai/api/v1", api_key=api_key)
    return client, model_name


def ask_openrouter(client, model_name, prompt):
    # This returns the assistant message from the first choice.
    response = client.chat.completions.create(
        model=model_name,
        messages=[{"role": "user", "content": prompt}],
    )
    return response.choices[0].message.content.strip()
```

```python
if MODEL_PROVIDER == "gemini":
    chat = start_gemini_chat(GOOGLE_API_KEY, GEMINI_MODEL)

    def ask_model(prompt):
        return ask_gemini(chat, prompt)

elif MODEL_PROVIDER == "openrouter":
    client, openrouter_model = start_openrouter_client(OPENROUTER_API_KEY, OPENROUTER_MODEL)

    def ask_model(prompt):
        return ask_openrouter(client, openrouter_model, prompt)

else:
    raise ValueError("MODEL_PROVIDER must be 'gemini' or 'openrouter'")
```
