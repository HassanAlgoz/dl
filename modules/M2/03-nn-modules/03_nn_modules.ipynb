```python
# imports
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
# model = nn.Sequential(
#     nn.Linear(1, 3)
#     nn.ReLU(),
#     nn.Linear(3, 1)
# )
```

```python
# Building a simple NN
class SimpleDNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleDNN, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.fc2 = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x
```

```python
# Initialize the model
input_size = 2
hidden_size = 4
output_size = 1
model = SimpleDNN(input_size, hidden_size, output_size)
print(model)
```

## Exercise 1:

Create a Network Modify the above network to include three hidden layers with ReLU activations.

Exercise 1: Modify Model Structure

1. Add a new hidden layer with 6 neurons to the SimpleDNN class.
2. Use ReLU activation for the new layer.
3. Print the updated model structure.

```python
# Solution for Exercise 1
class UpdatedDNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(UpdatedDNN, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.fc2 = nn.Linear(hidden_size, hidden_size)
        self.fc3 = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x

# Initialize the updated model
model = UpdatedDNN(input_size=2, hidden_size=6, output_size=1)
print(model)
```

---

We're going to create the following model for the next exercise, `MultiLayerDNN`

```python
class MultiLayerDNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(MultiLayerDNN, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.fc2 = nn.Linear(hidden_size, hidden_size)
        self.fc3 = nn.Linear(hidden_size, hidden_size)
        self.fc4 = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = F.relu(self.fc3(x))
        x = self.fc4(x)
        return x

# Initialize the new model
model = MultiLayerDNN(input_size, hidden_size, output_size)
print(model)
```
