# Detect language, then translate

The translation pipeline can be initialized with:
```python
pipeline("translation")
```

Helsinki-NLP models are **language-pair specific** (see the lab lesson). `opus-mt-es-en` only handles Spanish → English; another source language needs a different model.

**Your task:** Do not assume the source language. Detect it first, then load the matching source→English translation model.

**Hints:**
- Language identification is a text-classification problem — the same `pipeline` task you used for sentiment.
- Check the [language-detection model card](https://huggingface.co/papluca/xlm-roberta-base-language-detection) for its labels and intended use.
- Helsinki-NLP model IDs follow a `{src}` / `{tgt}` pattern. Browse models [here](https://huggingface.co/models?sort=trending&search=Helsinki-NLP).

Key translation parameters:
- `max_length`: Maximum length of generated translation
- `num_beams`: Number of beams for beam search (higher = more thorough but slower)
- `do_sample`: Whether to use sampling instead of beam search


````python
from transformers import pipeline

lang_pipe = pipeline(
    task="text-classification",
    model="papluca/xlm-roberta-base-language-detection"
  )
````

````python
text = [
    "Brevity is the soul of wit.",
    "من جد وجد ومن زرع حصد"
]
output = lang_pipe(text, top_k=1, truncation=True)
output
````

````python
# INSERT YOUR CODE HERE
# Detect the language of spanish_review, then translate it to English
lang_pipe(spanish_review)
````

**Hints:**
- Reuse the same detect-then-translate pattern you wrote for `spanish_review`.
- Browse Helsinki-NLP models [here](https://huggingface.co/models?sort=trending&search=Helsinki-NLP).

````python
# INSERT YOUR CODE HERE

source_language = lang
target_language = 'en'

trans_pipe = pipeline(
    task="translation",
    model=f"Helsinki-NLP/opus-mt-{source_language}-{target_language}"
)
````

````python
translated = trans_pipe(ikea_text)
print(translated)
````

````python
def auto_translate(text: str) -> str:
    lang = lang_pipe(text)[0]['label']
    source_language = lang
    target_language = 'en'
    trans_pipe = pipeline(
        task="translation",
        model=f"Helsinki-NLP/opus-mt-{source_language}-{target_language}"
    )
    translated = trans_pipe(text)
    return translated[0]['translation_text']
````

````python
trans1 = auto_translate(ikea_text)
trans2 = auto_translate(spanish_review)
````

## Conclusion
- Language detection is a text-classification step that tells you which bilingual translation model to load
- Helsinki-NLP `opus-mt-{src}-{tgt}` models are pair-specific — the wrong source language produces poor translations
- Seq2Seq models transform one sequence into another (translation, summarization)
- Translation and summarization can be chained for international product reviews
- Real-world applications often require combining multiple models
